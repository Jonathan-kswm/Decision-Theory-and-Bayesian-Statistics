# Bayesian Statistics

```python
import math
```

```python
# Prior
p_double_headed = 1/10          # P(A)
p_fair = 9/10                   # P(not A)

# Likelihoods
likelihood_double = 1 ** 4      # P(B | double-headed) = 1^4 = 1
likelihood_fair   = 0.5 ** 4     # P(B | fair) = 0.5^4

# Marginal probability of evidence
p_evidence = likelihood_double * p_double_headed + likelihood_fair * p_fair

# Posterior
posterior_double = likelihood_double * p_double_headed / p_evidence
posterior_fair   = likelihood_fair   * p_fair           / p_evidence

print(f"P(double-headed | 4 heads) = {posterior_double:.4f}")
print(f"P(fair | 4 heads)          = {posterior_fair:.4f}")
```

## Problem from PS2
- n = 20 iid Bernoulli trials
- P = $\theta$
- y = 16
- $\theta$ is uniform 0,1
- $P[0.5 < \theta < 0.57 | y]$

```python
import math
import numpy as np

# X ~ Bin(20 , theta) and see when X = 16
# set theta to randomly be between 0.5 and 0.75 
# run this a lot of times and see how many times X = 16 
# then no. 16 / no . runs = P[0.5 < theta < 0.75 | y] ???

# different thetas:
thetas = [0.5, 0.55, 0.6, 0.65, 0.7, 0.75]

def test(thetas=thetas, size=100000):
    out = {str(theta): 0 for theta in thetas}
    out["total"] = 0

    for theta in thetas:
        draws = np.random.binomial(20, theta, size)
        count = int(np.sum(draws == 16))
        out[str(theta)] = count / size
        out["total"] += count

    return out
```
